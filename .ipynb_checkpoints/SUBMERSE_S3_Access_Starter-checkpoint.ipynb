# SUBMERSE: Accessing DAS recordings through S3

This notebook demonstrates **access to** SUBMERSE DAS recordings stored in MinIO. It does not provide data analysis or visualisation workflows.

You will need an authorised S3 access key and secret key. **Do not save credentials in this notebook, share them, or commit them to Git.** The access policy is read-only.

## 1. Dependencies

Install the packages listed in `requirements.txt` in your own Python environment, then select that environment as your Jupyter kernel.

```bash
python -m pip install -r requirements.txt
```

The S3 endpoint uses HTTPS. Certificate verification remains enabled; if your environment has an unusual CA configuration, ask your system administrator rather than disabling TLS verification.

````python
import os
from getpass import getpass
import boto3
import s3fs
import h5py
import ssl
from botocore.config import Config

ENDPOINT = "https://s3.minio-z.submerse.eu"
BUCKET = "open-buffer"

# Optional: set MINIO_ACCESS_KEY_ID and MINIO_SECRET_ACCESS_KEY in your environment.
access_key = os.environ.get("MINIO_ACCESS_KEY") or getpass.getpass("MinIO access key: ")
secret_key = os.environ.get("MINIO_SECRET_KEY") or getpass.getpass("MinIO secret key: ")

# Use the operating system's trusted CA certificates by default.
# If necessary, set SSL_CERT_FILE to your trusted CA bundle path.
#ca_bundle = os.environ.get("SSL_CERT_FILE") or True

ca_bundle = os.environ.get("AWS_CA_BUNDLE")

if not ca_bundle:
    candidates = [
        ssl.get_default_verify_paths().cafile,
        "/etc/ssl/certs/ca-certificates.crt",
        "/etc/pki/tls/certs/ca-bundle.crt",
        "/etc/ssl/cert.pem",
    ]

    ca_bundle = next(
        (path for path in candidates
         if path and os.path.isfile(path)),
        True
    )

print("Certificate verification:", ca_bundle)

client = boto3.client(
    "s3",
    endpoint_url=ENDPOINT,
    aws_access_key_id=access_key,
    aws_secret_access_key=secret_key,
    region_name="us-east-1",
    verify=ca_bundle,
    config=Config(s3={"addressing_style": "path"}),
)

fs = s3fs.S3FileSystem(
    key=access_key,
    secret=secret_key,
    client_kwargs={
        "endpoint_url": ENDPOINT,
        "region_name": "us-east-1",
        "verify": ca_bundle,
    },
    config_kwargs={"s3": {"addressing_style": "path"}},
)
print("S3 clients configured. Credentials are not displayed.")
````

## 2. List available recordings

Objects are identified by **keys** (paths inside the bucket). This example lists up to 20 matching objects. Modify `PREFIX` to browse a different location or date.

````python
PREFIX = "AMS9SciencePark/AMS-LON1/57/GEANTDAS-001/20251201/"

response = client.list_objects_v2(Bucket=BUCKET, Prefix=PREFIX, MaxKeys=20)
keys = [obj["Key"] for obj in response.get("Contents", [])]

for key in keys:
    print(key)

if not keys:
    print("No objects found under this prefix. Check your access and prefix.")

# This is a demonstration key; replace it with a key you have permission to access.
selected_key = keys[0] if keys else None
````

## 3. Inspect HDF5 metadata remotely

The HDF5 file is opened through S3 without explicitly downloading it to local disk. Opening the file and reading metadata can still trigger S3 requests. The main data array is called `/data` in the tested example.

````python
if selected_key is None:
    raise ValueError("Select an available object key first.")

with fs.open(f"{BUCKET}/{selected_key}", "rb", block_size=1024 * 1024, cache_type="readahead") as remote:
    with h5py.File(remote, "r") as h5:
        dset = h5["data"]
        print("Shape:", dset.shape)
        print("Data type:", dset.dtype)
        print("HDF5 chunks:", dset.chunks)
        print("Compression:", dset.compression)
        print("Sampling interval (s):", h5["header/dt"][()])
        print("Channel spacing (m):", h5["header/dx"][()])
        print("Gauge length (m):", h5["header/gaugeLength"][()])
````

## 4. Read a small subset with buffering

The tested DAS files have a contiguous, uncompressed `/data` array organised as **time samples × fibre channels**. Narrow channel selections across many time samples can cause many small S3 range requests.

`cache_type="readahead"` with a **1 MiB block size** reduced the number of requests in our test. It also fetched more data. Use bounded selections to avoid large memory use and unnecessary transfers.

````python
# Example: 1,000 time samples and 100 channels.
# Change these indices to suit your work, keeping selections reasonably small.
time_start, time_stop = 0, 1000
channel_start, channel_stop = 0, 100

with fs.open(
    f"{BUCKET}/{selected_key}",
    "rb",
    block_size=1024 * 1024,
    cache_type="readahead",
) as remote:
    with h5py.File(remote, "r") as h5:
        data = h5["data"][time_start:time_stop, channel_start:channel_stop]

print("Loaded array shape:", data.shape)
print("Array size (MiB):", round(data.nbytes / 1024**2, 3))
````

## 5. What the buffering benchmark showed

For the **same 1,000 × 100 selection** on a tested recording:

| Remote access method | S3 GET requests | Requested byte ranges |
| --- | ---: | ---: |
| No caching (`cache_type="none"`) | 1,009 | 0.19 MiB |
| 1 MiB readahead (`cache_type="readahead"`) | 26 | 25.14 MiB |

Buffered access reduced the request count by about **97.4%**, while increasing bytes requested. The buffered test took **0.22 seconds** in the test environment. These are observations, **not guaranteed performance** for other networks, versions, selections or files.

The main `/data` dataset in the tested file was `(16000, 12500)`, `int16`, **contiguous** (`chunks=None`) and **uncompressed**. Different HDF5 layouts can behave differently.
