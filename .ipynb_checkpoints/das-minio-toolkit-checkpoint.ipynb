## 2. Connect using your S3 credentials
Use the access key and secret key provided for your group. The bucket remains private. The notebook uses path-style S3 addressing.

```python
import getpass
import os
import boto3
import s3fs
from botocore.config import Config

ENDPOINT = "https://s3.minio-z.submerse.eu"
BUCKET = "open-buffer"

access_key = os.environ.get("MINIO_ACCESS_KEY") or getpass.getpass("MinIO access key: ")
secret_key = os.environ.get("MINIO_SECRET_KEY") or getpass.getpass("MinIO secret key: ")

s3 = boto3.client(
    "s3", endpoint_url=ENDPOINT,
    aws_access_key_id=access_key, aws_secret_access_key=secret_key,
    region_name="us-east-1",
    config=Config(signature_version="s3v4", s3={"addressing_style": "path"}),
)
fs = s3fs.S3FileSystem(
    key=access_key, secret=secret_key,
    client_kwargs={"endpoint_url": ENDPOINT, "region_name": "us-east-1"},
    config_kwargs={"s3": {"addressing_style": "path"}},
    default_block_size=8 * 1024 * 1024,
)
print("Connected. Bucket listing follows:")
print([x["Key"] for x in s3.list_objects_v2(Bucket=BUCKET, Delimiter="/").get("Contents", [])])
print([x["Prefix"] for x in s3.list_objects_v2(Bucket=BUCKET, Delimiter="/").get("CommonPrefixes", [])])
```

## 4. Inspect one HDF5 object remotely

`h5py` requires a seekable file. `s3fs.open` provides a seekable, block-cached file-like object and requests ranges from MinIO. Opening an HDF5 object or inspecting metadata may trigger multiple S3 requests. Avoid listing enormous dataset contents.

```python
import h5py

if not recordings:
    raise ValueError("No matching recordings; change DATE or time filters")
selected_key = recordings[0]["key"]
print("Inspecting:", selected_key)

with fs.open(f"{BUCKET}/{selected_key}", "rb", block_size=8 * 1024 * 1024) as remote:
    with h5py.File(remote, "r") as h5:
        def describe(name, item):
            if isinstance(item, h5py.Dataset):
                print(f"{name}: shape={item.shape}, dtype={item.dtype}, chunks={item.chunks}")
        h5.visititems(describe)
```

## 5. Read a small slice of DAS data

The data path may differ between interrogators. Set `DATASET_PATH` after inspecting the output above. The example assumes the first axis is time and the second axis is channel; verify this for your instrument before analysing. Avoid `dataset[:]` on a large dataset.

```python
import numpy as np

DATASET_PATH = "data"  # CHANGE after inspecting the HDF5 structure
TIME_SLICE = slice(0, 1000)
CHANNEL_SLICE = slice(0, 100)

with fs.open(f"{BUCKET}/{selected_key}", "rb", block_size=8 * 1024 * 1024) as remote:
    with h5py.File(remote, "r") as h5:
        if DATASET_PATH not in h5:
            raise KeyError(f"Dataset {DATASET_PATH!r} not found. Check the inspection output above.")
        dataset = h5[DATASET_PATH]
        if dataset.ndim != 2:
            raise ValueError(f"Expected 2D DAS dataset, got {dataset.ndim} dimensions")
        sample = np.asarray(dataset[TIME_SLICE, CHANNEL_SLICE])

print("Selected array shape:", sample.shape)
print("Array memory:", round(sample.nbytes / 2**20, 2), "MiB")
```

## 6. Plot the selected samples
This is a raw array visualisation. Convert to calibrated units only after reading the relevant instrument metadata.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 4))
plt.imshow(sample.T, aspect="auto", origin="lower", interpolation="nearest")
plt.xlabel("Selected sample index")
plt.ylabel("Selected channel index")
plt.colorbar(label="Stored sample value")
plt.tight_layout()
plt.show()
```

## Notes and limitations

- **No full-file download is required by the notebook**, but HDF5 access can still transfer substantial amounts of data. Contiguous datasets may be relatively efficient for certain slices; chunked/compressed datasets may require downloading complete chunks that intersect a selection.
- `s3fs` caches byte ranges. Reopening a file or changing the access pattern can trigger additional requests. Large numbers of tiny reads can be slow over the internet.
- The HDF5 layout must be supported by `h5py`; some remote access patterns can be inefficient or unsupported by particular file structures.
- A successful read confirms access to the selected object; a separate permissions check is needed to establish that the account cannot write.
- Never share credentials or presigned URLs in public notebooks. The student's group policy should remain read-only.
- For heavy analysis of large time/channel regions, a purpose-built cloud-native representation (e.g., Zarr) or running compute close to MinIO may perform better.
